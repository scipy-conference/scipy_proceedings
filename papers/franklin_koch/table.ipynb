```python
#| label: tbl:kindetic-params

# Create the first DataFrame
df1 = pd.DataFrame({
    'A': ['A0', 'A1', 'A2'],
    'B': ['B0', 'B1', 'B2']
}, index=['K0', 'K1', 'K2'])

# Create the second DataFrame with different columns and index
df2 = pd.DataFrame({
    'C': ['C0', 'C1', 'C2'],
    'D': ['D0', 'D1', 'D2']
}, index=['K0', 'K2', 'K3'])

# Concatenate the DataFrames with a hierarchical index
df_concat = pd.concat([df1, df2], 
                      keys=['df1', 'df2'],  # Add keys for hierarchical indexing
                      axis=1,  # Concatenate along rows
                      join='outer', # Use 'outer' join to keep all columns
                      sort=False) # Avoid sorting the index

# Print the concatenated DataFrame
df_concat
```
