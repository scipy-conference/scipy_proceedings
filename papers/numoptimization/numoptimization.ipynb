```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.optimize import minimize
from numpy.linalg import cond
import time
import warnings

warnings.filterwarnings("ignore")

# ============================================================
# Settings
# ============================================================

SEEDS = list(range(20))
METHODS = ["BFGS", "L-BFGS-B", "Nelder-Mead", "Powell", "CG"]

# ============================================================
# Objective functions
# ============================================================

def rosenbrock(x):
    return (1 - x[0])**2 + 100 * (x[1] - x[0]**2)**2

def rastrigin(x):
    return 20 + x[0]**2 + x[1]**2 - 10 * (
        np.cos(2 * np.pi * x[0]) + np.cos(2 * np.pi * x[1])
    )

def badly_scaled(x):
    return (1e6 * x[0] - 1)**2 + (1e-6 * x[1] - 1)**2

def make_noisy_rosenbrock(sigma, rng):
    def f(x):
        return rosenbrock(x) + rng.normal(0, sigma)
    return f

def overflow_objective(x):
    val = np.exp(np.clip(x[0], -700, 700)) + (x[1] - 1)**2
    return float(val)

# ============================================================
# Random starting points
# ============================================================

def random_start(problem, rng):
    if problem == "Rosenbrock":
        return rng.uniform(-2, 2, size=2)
    if problem == "Rastrigin":
        return rng.uniform(-5.12, 5.12, size=2)
    if problem == "Badly Scaled":
        return rng.uniform(-2, 2, size=2)
    if problem == "Noisy Rosenbrock":
        return rng.uniform(-2, 2, size=2)
    if problem == "Overflow Objective":
        return rng.uniform(-5, 100, size=2)
    return rng.uniform(-2, 2, size=2)

# ============================================================
# Optimizer runner
# ============================================================

def run_optimizer(problem, func, x0, method, seed, maxiter=1000):
    history = []
    values = []

    def callback(xk):
        history.append(np.copy(xk))
        try:
            values.append(func(xk))
        except Exception:
            values.append(np.nan)

    start = time.time()

    try:
        result = minimize(
            func,
            x0,
            method=method,
            callback=callback,
            options={
                "maxiter": maxiter,
                "disp": False
            }
        )

        elapsed = time.time() - start

        final_value = float(result.fun) if np.isfinite(result.fun) else np.nan
        solution_norm = float(np.linalg.norm(result.x)) if np.all(np.isfinite(result.x)) else np.nan

        return {
            "problem": problem,
            "method": method,
            "seed": seed,
            "success": bool(result.success),
            "final_value": final_value,
            "iterations": getattr(result, "nit", np.nan),
            "function_evals": getattr(result, "nfev", np.nan),
            "time_seconds": elapsed,
            "solution_norm": solution_norm,
            "message": str(result.message),
            "history_length": len(history),
            "history": np.array(history),
            "values": np.array(values)
        }

    except Exception as e:
        elapsed = time.time() - start
        return {
            "problem": problem,
            "method": method,
            "seed": seed,
            "success": False,
            "final_value": np.nan,
            "iterations": np.nan,
            "function_evals": np.nan,
            "time_seconds": elapsed,
            "solution_norm": np.nan,
            "message": str(e),
            "history_length": len(history),
            "history": np.array(history),
            "values": np.array(values)
        }

# ============================================================
# Run multi-seed experiments
# ============================================================

raw_results = []

for seed in SEEDS:
    rng = np.random.default_rng(seed)

    problem_defs = [
        ("Rosenbrock", rosenbrock),
        ("Rastrigin", rastrigin),
        ("Badly Scaled", badly_scaled),
        ("Overflow Objective", overflow_objective),
    ]

    # add noisy objective separately so each seed gets its own RNG
    noise_rng = np.random.default_rng(seed + 1000)
    problem_defs.append(("Noisy Rosenbrock", make_noisy_rosenbrock(1e-2, noise_rng)))

    for problem_name, func in problem_defs:
        x0 = random_start(problem_name, rng)

        for method in METHODS:
            print(f"seed={seed:02d} | {problem_name:18s} | {method}")
            result = run_optimizer(problem_name, func, x0, method, seed)
            raw_results.append(result)

# ============================================================
# Convert to dataframe
# ============================================================

df = pd.DataFrame([
    {
        "problem": r["problem"],
        "method": r["method"],
        "seed": r["seed"],
        "success": r["success"],
        "final_value": r["final_value"],
        "iterations": r["iterations"],
        "function_evals": r["function_evals"],
        "time_seconds": r["time_seconds"],
        "solution_norm": r["solution_norm"],
        "history_length": r["history_length"],
        "message": r["message"][:120]
    }
    for r in raw_results
])

display(df.head())
df.to_csv("multi_run_raw_results.csv", index=False)

# ============================================================
# Summary statistics
# ============================================================

summary = df.groupby(["problem", "method"]).agg(
    runs=("seed", "count"),
    success_rate=("success", "mean"),
    mean_final_value=("final_value", "mean"),
    std_final_value=("final_value", "std"),
    median_final_value=("final_value", "median"),
    mean_iterations=("iterations", "mean"),
    std_iterations=("iterations", "std"),
    mean_function_evals=("function_evals", "mean"),
    mean_time_seconds=("time_seconds", "mean"),
).reset_index()

display(summary)
summary.to_csv("multi_run_summary.csv", index=False)

# ============================================================
# Ranking by problem
# ============================================================

ranked = summary.copy()
ranked["score"] = (
    ranked["success_rate"] * 100
    - np.log10(ranked["mean_final_value"].abs() + 1e-12)
    - 0.01 * ranked["mean_function_evals"]
)

ranked = ranked.sort_values(["problem", "score"], ascending=[True, False])
display(ranked)
ranked.to_csv("multi_run_ranked_optimizers.csv", index=False)

# ============================================================
# Plot success rates
# ============================================================

for problem in df["problem"].unique():
    temp = summary[summary["problem"] == problem]

    plt.figure(figsize=(8, 5))
    plt.bar(temp["method"], temp["success_rate"])
    plt.ylim(0, 1.05)
    plt.ylabel("Success Rate")
    plt.title(f"Success Rate Across Seeds: {problem}")
    plt.xticks(rotation=30)
    plt.grid(axis="y")
    plt.show()

# ============================================================
# Plot mean final objective values
# ============================================================

for problem in df["problem"].unique():
    temp = summary[summary["problem"] == problem]

    plt.figure(figsize=(8, 5))
    plt.bar(temp["method"], temp["mean_final_value"])
    plt.yscale("symlog", linthresh=1e-8)
    plt.ylabel("Mean Final Objective Value")
    plt.title(f"Mean Final Value Across Seeds: {problem}")
    plt.xticks(rotation=30)
    plt.grid(axis="y")
    plt.show()

# ============================================================
# Plot final value distribution
# ============================================================

for problem in df["problem"].unique():
    plt.figure(figsize=(8, 5))

    data = []
    labels = []

    for method in METHODS:
        vals = df[(df["problem"] == problem) & (df["method"] == method)]["final_value"].dropna().values
        if len(vals) > 0:
            data.append(vals)
            labels.append(method)

    plt.boxplot(data, labels=labels)
    plt.yscale("symlog", linthresh=1e-8)
    plt.ylabel("Final Objective Value")
    plt.title(f"Final Value Distribution: {problem}")
    plt.xticks(rotation=30)
    plt.grid(axis="y")
    plt.show()

# ============================================================
# Gradient norm approximation for Rosenbrock
# ============================================================

def numerical_grad(f, x, h=1e-6):
    x = np.asarray(x, dtype=float)
    grad = np.zeros_like(x)

    for i in range(len(x)):
        xp = x.copy()
        xm = x.copy()
        xp[i] += h
        xm[i] -= h
        grad[i] = (f(xp) - f(xm)) / (2 * h)

    return grad

grad_records = []

for r in raw_results:
    if r["problem"] == "Rosenbrock" and len(r["history"]) > 0:
        for t, x in enumerate(r["history"]):
            gnorm = np.linalg.norm(numerical_grad(rosenbrock, x))
            grad_records.append({
                "method": r["method"],
                "seed": r["seed"],
                "iteration": t,
                "gradient_norm": gnorm
            })

grad_df = pd.DataFrame(grad_records)
grad_df.to_csv("rosenbrock_gradient_norms.csv", index=False)

for method in METHODS:
    temp = grad_df[grad_df["method"] == method]

    if len(temp) == 0:
        continue

    mean_curve = temp.groupby("iteration")["gradient_norm"].mean().reset_index()

    plt.figure(figsize=(8, 5))
    plt.plot(mean_curve["iteration"], mean_curve["gradient_norm"])
    plt.yscale("log")
    plt.xlabel("Iteration")
    plt.ylabel("Mean Gradient Norm")
    plt.title(f"Mean Gradient Norm Evolution: {method} on Rosenbrock")
    plt.grid(True)
    plt.show()

# ============================================================
# Conditioning study with multiple sample sizes
# ============================================================

condition_records = []

for n_samples in [25, 50, 100, 200]:
    x = np.linspace(-1, 1, n_samples)

    for degree in range(2, 31):
        V = np.vander(x, degree)
        condition_records.append({
            "samples": n_samples,
            "degree": degree,
            "condition_number": cond(V)
        })

cond_df = pd.DataFrame(condition_records)
cond_df.to_csv("condition_number_multi_sample.csv", index=False)

plt.figure(figsize=(8, 5))

for n_samples in [25, 50, 100, 200]:
    temp = cond_df[cond_df["samples"] == n_samples]
    plt.semilogy(temp["degree"], temp["condition_number"], marker="o", label=f"n={n_samples}")

plt.xlabel("Polynomial Degree")
plt.ylabel("Condition Number")
plt.title("Condition Number Growth Across Sample Sizes")
plt.legend()
plt.grid(True)
plt.show()

print("Done.")
print("Saved files:")
print("- multi_run_raw_results.csv")
print("- multi_run_summary.csv")
print("- multi_run_ranked_optimizers.csv")
print("- rosenbrock_gradient_norms.csv")
print("- condition_number_multi_sample.csv")
```

```python
import numpy as np
import pandas as pd
import time
import warnings

from scipy.optimize import minimize
from scipy.special import expit
from numpy.linalg import norm, cond
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, log_loss

warnings.filterwarnings("ignore")

# ============================================================
# Real-data optimization experiment:
# Breast Cancer Wisconsin logistic regression via scipy.optimize
# ============================================================

SEEDS = list(range(100))
METHODS = ["BFGS", "L-BFGS-B", "CG", "Powell"]
L2 = 1e-3
MAXITER = 500

data = load_breast_cancer()
X_full = data.data
y_full = data.target.astype(float)

def add_intercept(X):
    return np.column_stack([np.ones(X.shape[0]), X])

def logistic_loss(w, X, y, l2=L2):
    z = X @ w
    # stable negative log-likelihood
    nll = np.mean(np.logaddexp(0, z) - y * z)
    reg = 0.5 * l2 * np.sum(w[1:] ** 2)
    return nll + reg

def logistic_grad(w, X, y, l2=L2):
    z = X @ w
    p = expit(z)
    grad = X.T @ (p - y) / X.shape[0]
    grad[1:] += l2 * w[1:]
    return grad

def hessian_condition(w, X, l2=L2):
    z = X @ w
    p = expit(z)
    s = p * (1 - p)
    H = (X.T * s) @ X / X.shape[0]
    H[1:, 1:] += l2 * np.eye(X.shape[1] - 1)
    try:
        return cond(H)
    except Exception:
        return np.inf

def run_single(seed, method, scaled):
    X_train, X_test, y_train, y_test = train_test_split(
        X_full,
        y_full,
        test_size=0.25,
        random_state=seed,
        stratify=y_full
    )

    if scaled:
        scaler = StandardScaler()
        X_train = scaler.fit_transform(X_train)
        X_test = scaler.transform(X_test)

    X_train_i = add_intercept(X_train)
    X_test_i = add_intercept(X_test)

    rng = np.random.default_rng(seed)
    w0 = rng.normal(0, 0.01, X_train_i.shape[1])

    start = time.time()

    use_jac = method != "Powell"

    result = minimize(
        logistic_loss,
        w0,
        args=(X_train_i, y_train),
        method=method,
        jac=logistic_grad if use_jac else None,
        options={
            "maxiter": MAXITER,
            "disp": False
        }
    )

    elapsed = time.time() - start

    w = result.x
    train_loss = logistic_loss(w, X_train_i, y_train)
    test_probs = expit(X_test_i @ w)
    test_preds = (test_probs >= 0.5).astype(int)

    test_logloss = log_loss(y_test, test_probs, labels=[0, 1])
    test_acc = accuracy_score(y_test, test_preds)

    grad_norm = norm(logistic_grad(w, X_train_i, y_train))
    hess_cond = hessian_condition(w, X_train_i)

    return {
        "seed": seed,
        "method": method,
        "scaled": scaled,
        "success": bool(result.success),
        "train_loss": train_loss,
        "test_logloss": test_logloss,
        "test_accuracy": test_acc,
        "iterations": getattr(result, "nit", np.nan),
        "function_evals": getattr(result, "nfev", np.nan),
        "gradient_norm": grad_norm,
        "hessian_condition": hess_cond,
        "time_seconds": elapsed,
        "message": str(result.message)[:100]
    }

# ============================================================
# Run experiment
# ============================================================

records = []

for seed in SEEDS:
    for scaled in [False, True]:
        for method in METHODS:
            print(f"seed={seed:02d} | scaled={scaled} | method={method}")
            records.append(run_single(seed, method, scaled))

df = pd.DataFrame(records)
display(df.head())

df.to_csv("real_data_logistic_optimization_raw.csv", index=False)

# ============================================================
# Summary table
# ============================================================

summary = df.groupby(["scaled", "method"]).agg(
    runs=("seed", "count"),
    success_rate=("success", "mean"),
    mean_test_logloss=("test_logloss", "mean"),
    std_test_logloss=("test_logloss", "std"),
    mean_test_accuracy=("test_accuracy", "mean"),
    std_test_accuracy=("test_accuracy", "std"),
    mean_train_loss=("train_loss", "mean"),
    mean_iterations=("iterations", "mean"),
    mean_function_evals=("function_evals", "mean"),
    mean_gradient_norm=("gradient_norm", "mean"),
    median_hessian_condition=("hessian_condition", "median"),
    mean_time_seconds=("time_seconds", "mean")
).reset_index()

summary["scaled"] = summary["scaled"].map({False: "Unscaled", True: "Scaled"})

display(summary)
summary.to_csv("real_data_logistic_optimization_summary.csv", index=False)

# ============================================================
# Stability comparison table
# ============================================================

stability = summary.copy()

stability["reliability_score"] = (
    100 * stability["success_rate"]
    + 10 * stability["mean_test_accuracy"]
    - stability["mean_test_logloss"]
    - np.log10(stability["median_hessian_condition"] + 1e-12)
    - 0.01 * stability["mean_function_evals"]
)

stability = stability.sort_values(
    by=["scaled", "reliability_score"],
    ascending=[True, False]
)

display(stability[[
    "scaled",
    "method",
    "success_rate",
    "mean_test_logloss",
    "mean_test_accuracy",
    "mean_gradient_norm",
    "median_hessian_condition",
    "mean_function_evals",
    "mean_time_seconds",
    "reliability_score"
]])

stability.to_csv("real_data_logistic_optimization_stability_ranking.csv", index=False)

# ============================================================
# Compact paper-ready table
# ============================================================

paper_table = summary[[
    "scaled",
    "method",
    "success_rate",
    "mean_test_logloss",
    "mean_test_accuracy",
    "mean_gradient_norm",
    "median_hessian_condition",
    "mean_function_evals",
    "mean_time_seconds"
]].copy()

paper_table.columns = [
    "Scaling",
    "Method",
    "Success Rate",
    "Mean Test LogLoss",
    "Mean Test Accuracy",
    "Mean Gradient Norm",
    "Median Hessian Condition",
    "Mean Function Evals",
    "Mean Runtime (s)"
]

display(paper_table)

print("Saved:")
print("- real_data_logistic_optimization_raw.csv")
print("- real_data_logistic_optimization_summary.csv")
print("- real_data_logistic_optimization_stability_ranking.csv")
```

```python
# ============================================================
# Real-data SciPy optimization experiment on UCI HIGGS
# Table-only output, no figures
# ============================================================

import os
import gzip
import time
import warnings
import urllib.request

import numpy as np
import pandas as pd

from scipy.optimize import minimize
from scipy.special import expit
from numpy.linalg import norm, cond

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, log_loss, roc_auc_score

warnings.filterwarnings("ignore")

# ============================================================
# Settings
# ============================================================

SEEDS = list(range(100))

# Keep this manageable for Colab.
# Increase N_ROWS to 100000 or 200000 if runtime is okay.
N_ROWS = 250000

TEST_SIZE = 0.25
L2 = 1e-3
MAXITER = 500

METHODS = ["BFGS", "L-BFGS-B", "CG", "Powell"]

DATA_URL = "https://archive.ics.uci.edu/ml/machine-learning-databases/00280/HIGGS.csv.gz"
DATA_PATH = "HIGGS.csv.gz"

# ============================================================
# Download dataset
# ============================================================

if not os.path.exists(DATA_PATH):
    print("Downloading HIGGS dataset. This file is large, about 2.6 GB compressed.")
    print("This may take time in Colab.")
    urllib.request.urlretrieve(DATA_URL, DATA_PATH)
else:
    print("HIGGS.csv.gz already exists.")

# ============================================================
# Load a subset
# ============================================================

print(f"Loading first {N_ROWS:,} rows from HIGGS...")

feature_names = [
    "lepton_pT", "lepton_eta", "lepton_phi",
    "missing_energy_magnitude", "missing_energy_phi",
    "jet_1_pt", "jet_1_eta", "jet_1_phi", "jet_1_btag",
    "jet_2_pt", "jet_2_eta", "jet_2_phi", "jet_2_btag",
    "jet_3_pt", "jet_3_eta", "jet_3_phi", "jet_3_btag",
    "jet_4_pt", "jet_4_eta", "jet_4_phi", "jet_4_btag",
    "m_jj", "m_jjj", "m_lv", "m_jlv",
    "m_bb", "m_wbb", "m_wwbb"
]

columns = ["target"] + feature_names

df = pd.read_csv(
    DATA_PATH,
    compression="gzip",
    header=None,
    names=columns,
    nrows=N_ROWS
)

X_full = df[feature_names].values.astype(np.float64)
y_full = df["target"].values.astype(np.float64)

print("Loaded shape:", X_full.shape)
print("Class balance:", pd.Series(y_full).value_counts(normalize=True).to_dict())

# ============================================================
# Logistic regression objective
# ============================================================

def add_intercept(X):
    return np.column_stack([np.ones(X.shape[0]), X])

def logistic_loss(w, X, y, l2=L2):
    z = X @ w
    nll = np.mean(np.logaddexp(0, z) - y * z)
    reg = 0.5 * l2 * np.sum(w[1:] ** 2)
    return float(nll + reg)

def logistic_grad(w, X, y, l2=L2):
    z = X @ w
    p = expit(z)
    grad = X.T @ (p - y) / X.shape[0]
    grad[1:] += l2 * w[1:]
    return grad

def hessian_condition(w, X, l2=L2):
    z = X @ w
    p = expit(z)
    s = p * (1 - p)

    H = (X.T * s) @ X / X.shape[0]
    H[1:, 1:] += l2 * np.eye(X.shape[1] - 1)

    try:
        return float(cond(H))
    except Exception:
        return np.inf

# ============================================================
# Run one optimization
# ============================================================

def run_single(seed, method, scaled):
    X_train, X_test, y_train, y_test = train_test_split(
        X_full,
        y_full,
        test_size=TEST_SIZE,
        random_state=seed,
        stratify=y_full
    )

    if scaled:
        scaler = StandardScaler()
        X_train = scaler.fit_transform(X_train)
        X_test = scaler.transform(X_test)

    X_train_i = add_intercept(X_train)
    X_test_i = add_intercept(X_test)

    rng = np.random.default_rng(seed)
    w0 = rng.normal(0, 0.01, X_train_i.shape[1])

    use_jac = method != "Powell"

    start = time.time()

    result = minimize(
        logistic_loss,
        w0,
        args=(X_train_i, y_train),
        method=method,
        jac=logistic_grad if use_jac else None,
        options={
            "maxiter": MAXITER,
            "disp": False
        }
    )

    elapsed = time.time() - start

    w = result.x

    train_loss = logistic_loss(w, X_train_i, y_train)
    train_grad_norm = norm(logistic_grad(w, X_train_i, y_train))
    hess_cond = hessian_condition(w, X_train_i)

    test_probs = expit(X_test_i @ w)
    test_preds = (test_probs >= 0.5).astype(int)

    test_logloss = log_loss(y_test, test_probs, labels=[0, 1])
    test_accuracy = accuracy_score(y_test, test_preds)
    test_auc = roc_auc_score(y_test, test_probs)

    return {
        "seed": seed,
        "method": method,
        "scaled": scaled,
        "success": bool(result.success),
        "train_loss": train_loss,
        "test_logloss": test_logloss,
        "test_accuracy": test_accuracy,
        "test_auc": test_auc,
        "iterations": getattr(result, "nit", np.nan),
        "function_evals": getattr(result, "nfev", np.nan),
        "gradient_norm": train_grad_norm,
        "hessian_condition": hess_cond,
        "time_seconds": elapsed,
        "message": str(result.message)[:120]
    }

# ============================================================
# Run full experiment
# ============================================================

records = []

total_runs = len(SEEDS) * 2 * len(METHODS)
run_count = 0

for seed in SEEDS:
    for scaled in [False, True]:
        for method in METHODS:
            run_count += 1
            print(f"[{run_count}/{total_runs}] seed={seed:03d} | scaled={scaled} | method={method}")
            try:
                records.append(run_single(seed, method, scaled))
            except Exception as e:
                records.append({
                    "seed": seed,
                    "method": method,
                    "scaled": scaled,
                    "success": False,
                    "train_loss": np.nan,
                    "test_logloss": np.nan,
                    "test_accuracy": np.nan,
                    "test_auc": np.nan,
                    "iterations": np.nan,
                    "function_evals": np.nan,
                    "gradient_norm": np.nan,
                    "hessian_condition": np.nan,
                    "time_seconds": np.nan,
                    "message": str(e)[:120]
                })

raw = pd.DataFrame(records)
raw.to_csv("higgs_scipy_optimization_raw_100seeds.csv", index=False)

display(raw.head())

# ============================================================
# Summary table
# ============================================================

summary = raw.groupby(["scaled", "method"]).agg(
    runs=("seed", "count"),
    success_rate=("success", "mean"),

    mean_test_logloss=("test_logloss", "mean"),
    std_test_logloss=("test_logloss", "std"),
    median_test_logloss=("test_logloss", "median"),

    mean_test_accuracy=("test_accuracy", "mean"),
    std_test_accuracy=("test_accuracy", "std"),

    mean_test_auc=("test_auc", "mean"),
    std_test_auc=("test_auc", "std"),

    mean_train_loss=("train_loss", "mean"),
    mean_gradient_norm=("gradient_norm", "mean"),
    median_gradient_norm=("gradient_norm", "median"),

    median_hessian_condition=("hessian_condition", "median"),
    mean_hessian_condition=("hessian_condition", "mean"),

    mean_iterations=("iterations", "mean"),
    mean_function_evals=("function_evals", "mean"),
    mean_time_seconds=("time_seconds", "mean")
).reset_index()

summary["scaled"] = summary["scaled"].map({False: "Unscaled", True: "Scaled"})

summary.to_csv("higgs_scipy_optimization_summary_100seeds.csv", index=False)

display(summary)

# ============================================================
# Paper-ready table
# ============================================================

paper_table = summary[[
    "scaled",
    "method",
    "success_rate",
    "mean_test_logloss",
    "std_test_logloss",
    "mean_test_accuracy",
    "std_test_accuracy",
    "mean_test_auc",
    "std_test_auc",
    "mean_gradient_norm",
    "median_hessian_condition",
    "mean_function_evals",
    "mean_time_seconds"
]].copy()

paper_table.columns = [
    "Scaling",
    "Method",
    "Success Rate",
    "Mean Test LogLoss",
    "Std Test LogLoss",
    "Mean Test Accuracy",
    "Std Test Accuracy",
    "Mean Test AUC",
    "Std Test AUC",
    "Mean Gradient Norm",
    "Median Hessian Condition",
    "Mean Function Evals",
    "Mean Runtime (s)"
]

paper_table.to_csv("higgs_paper_ready_table_100seeds.csv", index=False)

display(paper_table)

# ============================================================
# Optional compact LaTeX table output
# ============================================================

latex_table = paper_table.to_latex(
    index=False,
    float_format=lambda x: f"{x:.3e}" if abs(x) < 1e-2 or abs(x) > 1e3 else f"{x:.4f}",
    caption="SciPy optimizer reliability on the UCI HIGGS logistic-regression benchmark across 100 randomized train/test splits.",
    label="tab:higgs_real_data"
)

with open("higgs_paper_ready_table_100seeds.tex", "w") as f:
    f.write(latex_table)

print("Saved files:")
print("- higgs_scipy_optimization_raw_100seeds.csv")
print("- higgs_scipy_optimization_summary_100seeds.csv")
print("- higgs_paper_ready_table_100seeds.csv")
print("- higgs_paper_ready_table_100seeds.tex")
```

```python
import os
import gzip
import time
import warnings
import urllib.request

import numpy as np
import pandas as pd

from scipy.optimize import minimize
from scipy.special import expit
from numpy.linalg import norm, cond

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, log_loss, roc_auc_score

warnings.filterwarnings("ignore")

# ============================================================
# Settings
# ============================================================

SEEDS = list(range(100))

# Keep this manageable for Colab.
# Increase N_ROWS to 100000 or 200000 if runtime is okay.
N_ROWS = None

TEST_SIZE = 0.25
L2 = 1e-3
MAXITER = 500

METHODS = ["BFGS", "L-BFGS-B", "CG", "Powell"]

DATA_URL = "https://archive.ics.uci.edu/ml/machine-learning-databases/00280/HIGGS.csv.gz"
DATA_PATH = "HIGGS.csv.gz"

# ============================================================
# Download dataset
# ============================================================

if not os.path.exists(DATA_PATH):
    print("Downloading HIGGS dataset. This file is large, about 2.6 GB compressed.")
    print("This may take time in Colab.")
    urllib.request.urlretrieve(DATA_URL, DATA_PATH)
else:
    print("HIGGS.csv.gz already exists.")

# ============================================================
# Load a subset
# ============================================================

if N_ROWS is None:
    print("Loading all rows from HIGGS...")
else:
    print(f"Loading first {N_ROWS:,} rows from HIGGS...")

feature_names = [
    "lepton_pT", "lepton_eta", "lepton_phi",
    "missing_energy_magnitude", "missing_energy_phi",
    "jet_1_pt", "jet_1_eta", "jet_1_phi", "jet_1_btag",
    "jet_2_pt", "jet_2_eta", "jet_2_phi", "jet_2_btag",
    "jet_3_pt", "jet_3_eta", "jet_3_phi", "jet_3_btag",
    "jet_4_pt", "jet_4_eta", "jet_4_phi", "jet_4_btag",
    "m_jj", "m_jjj", "m_lv", "m_jlv",
    "m_bb", "m_wbb", "m_wwbb"
]

columns = ["target"] + feature_names

df = pd.read_csv(
    DATA_PATH,
    compression="gzip",
    header=None,
    names=columns,
    nrows=N_ROWS
)

X_full = df[feature_names].values.astype(np.float64)
y_full = df["target"].values.astype(np.float64)

print("Loaded shape:", X_full.shape)
print("Class balance:", pd.Series(y_full).value_counts(normalize=True).to_dict())

# ============================================================
# Logistic regression objective
# ============================================================

def add_intercept(X):
    return np.column_stack([np.ones(X.shape[0]), X])

def logistic_loss(w, X, y, l2=L2):
    z = X @ w
    nll = np.mean(np.logaddexp(0, z) - y * z)
    reg = 0.5 * l2 * np.sum(w[1:] ** 2)
    return float(nll + reg)

def logistic_grad(w, X, y, l2=L2):
    z = X @ w
    p = expit(z)
    grad = X.T @ (p - y) / X.shape[0]
    grad[1:] += l2 * w[1:]
    return grad

def hessian_condition(w, X, l2=L2):
    z = X @ w
    p = expit(z)
    s = p * (1 - p)

    H = (X.T * s) @ X / X.shape[0]
    H[1:, 1:] += l2 * np.eye(X.shape[1] - 1)

    try:
        return float(cond(H))
    except Exception:
        return np.inf

# ============================================================
# Run one optimization
# ============================================================

def run_single(seed, method, scaled):
    X_train, X_test, y_train, y_test = train_test_split(
        X_full,
        y_full,
        test_size=TEST_SIZE,
        random_state=seed,
        stratify=y_full
    )

    if scaled:
        scaler = StandardScaler()
        X_train = scaler.fit_transform(X_train)
        X_test = scaler.transform(X_test)

    X_train_i = add_intercept(X_train)
    X_test_i = add_intercept(X_test)

    rng = np.random.default_rng(seed)
    w0 = rng.normal(0, 0.01, X_train_i.shape[1])

    use_jac = method != "Powell"

    start = time.time()

    result = minimize(
        logistic_loss,
        w0,
        args=(X_train_i, y_train),
        method=method,
        jac=logistic_grad if use_jac else None,
        options={
            "maxiter": MAXITER,
            "disp": False
        }
    )

    elapsed = time.time() - start

    w = result.x

    train_loss = logistic_loss(w, X_train_i, y_train)
    train_grad_norm = norm(logistic_grad(w, X_train_i, y_train))
    hess_cond = hessian_condition(w, X_train_i)

    test_probs = expit(X_test_i @ w)
    test_preds = (test_probs >= 0.5).astype(int)

    test_logloss = log_loss(y_test, test_probs, labels=[0, 1])
    test_accuracy = accuracy_score(y_test, test_preds)
    test_auc = roc_auc_score(y_test, test_probs)

    return {
        "seed": seed,
        "method": method,
        "scaled": scaled,
        "success": bool(result.success),
        "train_loss": train_loss,
        "test_logloss": test_logloss,
        "test_accuracy": test_accuracy,
        "test_auc": test_auc,
        "iterations": getattr(result, "nit", np.nan),
        "function_evals": getattr(result, "nfev", np.nan),
        "gradient_norm": train_grad_norm,
        "hessian_condition": hess_cond,
        "time_seconds": elapsed,
        "message": str(result.message)[:120]
    }

# ============================================================
# Run full experiment
# ============================================================

records = []

total_runs = len(SEEDS) * 2 * len(METHODS)
run_count = 0

for seed in SEEDS:
    for scaled in [False, True]:
        for method in METHODS:
            run_count += 1
            print(f"[{run_count}/{total_runs}] seed={seed:03d} | scaled={scaled} | method={method}")
            try:
                records.append(run_single(seed, method, scaled))
            except Exception as e:
                records.append({
                    "seed": seed,
                    "method": method,
                    "scaled": scaled,
                    "success": False,
                    "train_loss": np.nan,
                    "test_logloss": np.nan,
                    "test_accuracy": np.nan,
                    "test_auc": np.nan,
                    "iterations": np.nan,
                    "function_evals": np.nan,
                    "gradient_norm": np.nan,
                    "hessian_condition": np.nan,
                    "time_seconds": np.nan,
                    "message": str(e)[:120]
                })

raw = pd.DataFrame(records)
raw.to_csv("higgs_scipy_optimization_raw_100seeds.csv", index=False)

display(raw.head())

# ============================================================
# Summary table
# ============================================================

summary = raw.groupby(["scaled", "method"]).agg(
    runs=("seed", "count"),
    success_rate=("success", "mean"),

    mean_test_logloss=("test_logloss", "mean"),
    std_test_logloss=("test_logloss", "std"),
    median_test_logloss=("test_logloss", "median"),

    mean_test_accuracy=("test_accuracy", "mean"),
    std_test_accuracy=("test_accuracy", "std"),

    mean_test_auc=("test_auc", "mean"),
    std_test_auc=("test_auc", "std"),

    mean_train_loss=("train_loss", "mean"),
    mean_gradient_norm=("gradient_norm", "mean"),
    median_gradient_norm=("gradient_norm", "median"),

    median_hessian_condition=("hessian_condition", "median"),
    mean_hessian_condition=("hessian_condition", "mean"),

    mean_iterations=("iterations", "mean"),
    mean_function_evals=("function_evals", "mean"),
    mean_time_seconds=("time_seconds", "mean")
).reset_index()

summary["scaled"] = summary["scaled"].map({False: "Unscaled", True: "Scaled"})

summary.to_csv("higgs_scipy_optimization_summary_100seeds.csv", index=False)

display(summary)

# ============================================================
# Paper-ready table
# ============================================================

paper_table = summary[[
    "scaled",
    "method",
    "success_rate",
    "mean_test_logloss",
    "std_test_logloss",
    "mean_test_accuracy",
    "std_test_accuracy",
    "mean_test_auc",
    "std_test_auc",
    "mean_gradient_norm",
    "median_hessian_condition",
    "mean_function_evals",
    "mean_time_seconds"
]].copy()

paper_table.columns = [
    "Scaling",
    "Method",
    "Success Rate",
    "Mean Test LogLoss",
    "Std Test LogLoss",
    "Mean Test Accuracy",
    "Std Test Accuracy",
    "Mean Test AUC",
    "Std Test AUC",
    "Mean Gradient Norm",
    "Median Hessian Condition",
    "Mean Function Evals",
    "Mean Runtime (s)"
]

paper_table.to_csv("higgs_paper_ready_table_100seeds.csv", index=False)

display(paper_table)

# ============================================================
# Optional compact LaTeX table output
# ============================================================

latex_table = paper_table.to_latex(
    index=False,
    float_format=lambda x: f"{x:.3e}" if abs(x) < 1e-2 or abs(x) > 1e3 else f"{x:.4f}",
    caption="SciPy optimizer reliability on the UCI HIGGS logistic-regression benchmark across 100 randomized train/test splits.",
    label="tab:higgs_real_data"
)

with open("higgs_paper_ready_table_100seeds.tex", "w") as f:
    f.write(latex_table)

print("Saved files:")
print("- higgs_scipy_optimization_raw_100seeds.csv")
print("- higgs_scipy_optimization_summary_100seeds.csv")
print("- higgs_paper_ready_table_100seeds.csv")
print("- higgs_paper_ready_table_100seeds.tex")
```

```python
# ============================================================
# HIGGS Instability Stress Test
# Controlled numerical instability on real scientific data
# ============================================================

import os
import gzip
import time
import warnings
import urllib.request

import numpy as np
import pandas as pd

from scipy.optimize import minimize
from scipy.special import expit
from numpy.linalg import norm, cond

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, log_loss

warnings.filterwarnings("ignore")

# ============================================================
# Settings
# ============================================================

SEEDS = list(range(50))

N_ROWS = 250000
TEST_SIZE = 0.25

METHODS = ["BFGS", "L-BFGS-B", "CG", "Powell"]

MAXITER = 500
L2 = 1e-3

DATA_URL = "https://archive.ics.uci.edu/ml/machine-learning-databases/00280/HIGGS.csv.gz"
DATA_PATH = "HIGGS.csv.gz"

# ============================================================
# Download dataset
# ============================================================

if not os.path.exists(DATA_PATH):
    urllib.request.urlretrieve(DATA_URL, DATA_PATH)

# ============================================================
# Load HIGGS subset
# ============================================================

columns = ["target"] + [f"f{i}" for i in range(28)]

df = pd.read_csv(
    DATA_PATH,
    compression="gzip",
    header=None,
    names=columns,
    nrows=N_ROWS
)

X_full = df.iloc[:, 1:].values.astype(np.float64)
y_full = df.iloc[:, 0].values.astype(np.float64)

print("Loaded:", X_full.shape)

# ============================================================
# Create instability
# ============================================================

def create_instability(X, rng):

    X_bad = X.copy()

    # --------------------------------------------------------
    # 1. Extreme scaling imbalance
    # --------------------------------------------------------

    scales = np.geomspace(1e-8, 1e8, X_bad.shape[1])
    X_bad *= scales

    # --------------------------------------------------------
    # 2. Near-collinear features
    # --------------------------------------------------------

    X_bad[:, 1] = X_bad[:, 0] * 0.999999
    X_bad[:, 2] = X_bad[:, 0] * 1.000001

    # --------------------------------------------------------
    # 3. Noise injection
    # --------------------------------------------------------

    noise = rng.normal(0, 1e-3, X_bad.shape)
    X_bad += noise

    return X_bad

# ============================================================
# Logistic objective
# ============================================================

def add_intercept(X):
    return np.column_stack([np.ones(X.shape[0]), X])

def logistic_loss(w, X, y, l2=L2):

    z = X @ w

    # intentionally unstable region possible
    probs = expit(z)

    eps = 1e-12

    loss = -np.mean(
        y * np.log(probs + eps)
        + (1 - y) * np.log(1 - probs + eps)
    )

    reg = 0.5 * l2 * np.sum(w[1:] ** 2)

    return loss + reg

def logistic_grad(w, X, y, l2=L2):

    z = X @ w
    probs = expit(z)

    grad = X.T @ (probs - y) / X.shape[0]
    grad[1:] += l2 * w[1:]

    return grad

# ============================================================
# Hessian condition
# ============================================================

def hessian_condition(w, X):

    z = X @ w
    p = expit(z)
    s = p * (1 - p)

    H = (X.T * s) @ X / X.shape[0]

    try:
        return cond(H)
    except:
        return np.inf

# ============================================================
# Main experiment
# ============================================================

records = []

total_runs = len(SEEDS) * 2 * len(METHODS)
run_count = 0

for seed in SEEDS:

    rng = np.random.default_rng(seed)

    X_train, X_test, y_train, y_test = train_test_split(
        X_full,
        y_full,
        test_size=TEST_SIZE,
        random_state=seed,
        stratify=y_full
    )

    # --------------------------------------------------------
    # Create instability
    # --------------------------------------------------------

    X_bad_train = create_instability(X_train, rng)
    X_bad_test = create_instability(X_test, rng)

    # --------------------------------------------------------
    # Stable version
    # --------------------------------------------------------

    scaler = StandardScaler()

    X_scaled_train = scaler.fit_transform(X_bad_train)
    X_scaled_test = scaler.transform(X_bad_test)

    variants = [
        ("Unstable", X_bad_train, X_bad_test),
        ("Scaled", X_scaled_train, X_scaled_test)
    ]

    for variant_name, Xtr, Xte in variants:

        Xtr_i = add_intercept(Xtr)
        Xte_i = add_intercept(Xte)

        for method in METHODS:

            run_count += 1

            print(
                f"[{run_count}/{total_runs}] "
                f"seed={seed} | {variant_name} | {method}"
            )

            x0 = rng.normal(0, 0.1, Xtr_i.shape[1])

            start = time.time()

            use_jac = method != "Powell"

            try:

                result = minimize(
                    logistic_loss,
                    x0,
                    args=(Xtr_i, y_train),
                    method=method,
                    jac=logistic_grad if use_jac else None,
                    options={
                        "maxiter": MAXITER,
                        "disp": False
                    }
                )

                elapsed = time.time() - start

                w = result.x

                probs = expit(Xte_i @ w)
                preds = (probs >= 0.5).astype(int)

                test_loss = log_loss(y_test, probs)
                test_acc = accuracy_score(y_test, preds)

                grad_norm = norm(
                    logistic_grad(w, Xtr_i, y_train)
                )

                hess_cond = hessian_condition(w, Xtr_i)

                # ------------------------------------------------
                # Practical success
                # ------------------------------------------------

                practical_success = (
                    result.success
                    and np.isfinite(test_loss)
                    and grad_norm < 1e-3
                    and hess_cond < 1e12
                )

                false_success = (
                    result.success
                    and not practical_success
                )

                records.append({
                    "seed": seed,
                    "variant": variant_name,
                    "method": method,

                    "optimizer_success": result.success,
                    "practical_success": practical_success,
                    "false_success": false_success,

                    "test_logloss": test_loss,
                    "test_accuracy": test_acc,

                    "gradient_norm": grad_norm,
                    "hessian_condition": hess_cond,

                    "iterations": getattr(result, "nit", np.nan),
                    "function_evals": getattr(result, "nfev", np.nan),

                    "runtime_seconds": elapsed,
                    "message": str(result.message)[:120]
                })

            except Exception as e:

                records.append({
                    "seed": seed,
                    "variant": variant_name,
                    "method": method,

                    "optimizer_success": False,
                    "practical_success": False,
                    "false_success": False,

                    "test_logloss": np.nan,
                    "test_accuracy": np.nan,

                    "gradient_norm": np.nan,
                    "hessian_condition": np.nan,

                    "iterations": np.nan,
                    "function_evals": np.nan,

                    "runtime_seconds": np.nan,
                    "message": str(e)
                })

# ============================================================
# Tables
# ============================================================

raw = pd.DataFrame(records)

raw.to_csv("higgs_instability_raw.csv", index=False)

summary = raw.groupby([
    "variant",
    "method"
]).agg(

    optimizer_success_rate=("optimizer_success", "mean"),
    practical_success_rate=("practical_success", "mean"),
    false_success_rate=("false_success", "mean"),

    mean_logloss=("test_logloss", "mean"),
    std_logloss=("test_logloss", "std"),

    mean_accuracy=("test_accuracy", "mean"),

    mean_gradient_norm=("gradient_norm", "mean"),

    median_hessian_condition=("hessian_condition", "median"),

    mean_function_evals=("function_evals", "mean"),

    mean_runtime=("runtime_seconds", "mean")

).reset_index()

display(summary)

summary.to_csv(
    "higgs_instability_summary.csv",
    index=False
)

print("Saved:")
print("- higgs_instability_raw.csv")
print("- higgs_instability_summary.csv")
```

```python
import numpy as np
import pandas as pd

DEGREES = [5, 10, 15, 20]
SEEDS = range(100)
N_SAMPLES = 100

records = []

for seed in SEEDS:
    rng = np.random.default_rng(seed)

    # random sample locations in [0, 1]
    x = np.sort(rng.uniform(0, 1, N_SAMPLES))

    for degree in DEGREES:
        # degree d means columns: 1, x, x^2, ..., x^d
        V = np.vander(x, N=degree + 1, increasing=True)

        cond_num = np.linalg.cond(V)

        records.append({
            "seed": seed,
            "polynomial_degree": degree,
            "condition_number": cond_num
        })

raw_df = pd.DataFrame(records)

summary_df = raw_df.groupby("polynomial_degree").agg(
    mean_condition_number=("condition_number", "mean"),
    std_condition_number=("condition_number", "std"),
    median_condition_number=("condition_number", "median"),
    min_condition_number=("condition_number", "min"),
    max_condition_number=("condition_number", "max")
).reset_index()

print(summary_df)

print("\nLaTeX Table:\n")
print(summary_df.to_latex(
    index=False,
    float_format=lambda x: f"{x:.3e}"
))

raw_df.to_csv("vandermonde_conditioning_raw_100seeds.csv", index=False)
summary_df.to_csv("vandermonde_conditioning_summary_100seeds.csv", index=False)
```

```python
# ============================================================
# HIGGS Instability Severity Ablation
# 100 seeds, table-only output
# ============================================================

import os
import time
import warnings
import urllib.request

import numpy as np
import pandas as pd

from scipy.optimize import minimize
from scipy.special import expit
from numpy.linalg import norm, cond

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, log_loss

warnings.filterwarnings("ignore")

# ============================================================
# Settings
# ============================================================

SEEDS = list(range(100))
N_ROWS = 250000
TEST_SIZE = 0.25

# Keep this focused and fast enough.
# These are the main scalable SciPy optimizers.
METHODS = ["BFGS", "L-BFGS-B", "CG"]

MAXITER = 500
L2 = 1e-3

DATA_URL = "https://archive.ics.uci.edu/ml/machine-learning-databases/00280/HIGGS.csv.gz"
DATA_PATH = "HIGGS.csv.gz"

# Severity levels:
# scale_exp = 2 means features scaled between 1e-2 and 1e2
# scale_exp = 4 means features scaled between 1e-4 and 1e4
# scale_exp = 8 means features scaled between 1e-8 and 1e8

SEVERITY_LEVELS = [
    {"severity": "Mild", "scale_exp": 2, "noise_std": 1e-3, "collinear": True},
    {"severity": "Moderate", "scale_exp": 4, "noise_std": 1e-3, "collinear": True},
    {"severity": "Extreme", "scale_exp": 8, "noise_std": 1e-3, "collinear": True},
]

# Practical success threshold used in the paper
GRADIENT_THRESHOLD = 1e-3
CONDITION_THRESHOLD = 1e12

# ============================================================
# Download HIGGS
# ============================================================

if not os.path.exists(DATA_PATH):
    print("Downloading HIGGS dataset. This is large.")
    urllib.request.urlretrieve(DATA_URL, DATA_PATH)
else:
    print("HIGGS.csv.gz already exists.")

# ============================================================
# Load HIGGS subset
# ============================================================

columns = ["target"] + [f"f{i}" for i in range(28)]

df = pd.read_csv(
    DATA_PATH,
    compression="gzip",
    header=None,
    names=columns,
    nrows=N_ROWS
)

X_full = df.iloc[:, 1:].values.astype(np.float64)
y_full = df.iloc[:, 0].values.astype(np.float64)

print("Loaded:", X_full.shape)

# ============================================================
# Helpers
# ============================================================

def add_intercept(X):
    return np.column_stack([np.ones(X.shape[0]), X])


def create_instability(X, rng, scale_exp, noise_std=1e-3, collinear=True):
    """
    Creates controlled instability using:
    1. scale imbalance from 1e-scale_exp to 1e+scale_exp
    2. near-collinearity among selected features
    3. additive Gaussian perturbation noise
    """
    X_bad = X.copy().astype(np.float64)

    # 1. Controlled feature-scale imbalance
    scales = np.geomspace(10.0 ** (-scale_exp), 10.0 ** scale_exp, X_bad.shape[1])
    X_bad *= scales

    # 2. Near-collinear feature construction
    if collinear and X_bad.shape[1] >= 3:
        X_bad[:, 1] = X_bad[:, 0] * 0.999999
        X_bad[:, 2] = X_bad[:, 0] * 1.000001

    # 3. Additive noise
    if noise_std > 0:
        X_bad += rng.normal(0, noise_std, X_bad.shape)

    return X_bad


def logistic_loss(w, X, y, l2=L2):
    z = X @ w
    nll = np.mean(np.logaddexp(0, z) - y * z)
    reg = 0.5 * l2 * np.sum(w[1:] ** 2)
    return float(nll + reg)


def logistic_grad(w, X, y, l2=L2):
    z = X @ w
    p = expit(z)

    grad = X.T @ (p - y) / X.shape[0]
    grad[1:] += l2 * w[1:]

    return grad


def hessian_condition(w, X, l2=L2):
    z = X @ w
    p = expit(z)
    s = p * (1 - p)

    H = (X.T * s) @ X / X.shape[0]
    H[1:, 1:] += l2 * np.eye(X.shape[1] - 1)

    try:
        return float(cond(H))
    except Exception:
        return np.inf


def run_one(seed, severity_name, scale_exp, method):
    rng = np.random.default_rng(seed)

    X_train, X_test, y_train, y_test = train_test_split(
        X_full,
        y_full,
        test_size=TEST_SIZE,
        random_state=seed,
        stratify=y_full
    )

    # Create unstable data
    X_bad_train = create_instability(
        X_train,
        rng,
        scale_exp=scale_exp,
        noise_std=1e-3,
        collinear=True
    )

    X_bad_test = create_instability(
        X_test,
        rng,
        scale_exp=scale_exp,
        noise_std=1e-3,
        collinear=True
    )

    Xtr_i = add_intercept(X_bad_train)
    Xte_i = add_intercept(X_bad_test)

    x0 = rng.normal(0, 0.1, Xtr_i.shape[1])

    start = time.time()

    try:
        result = minimize(
            logistic_loss,
            x0,
            args=(Xtr_i, y_train),
            method=method,
            jac=logistic_grad if method != "Powell" else None,
            options={
                "maxiter": MAXITER,
                "disp": False
            }
        )

        runtime = time.time() - start

        w = result.x

        test_probs = expit(Xte_i @ w)
        test_preds = (test_probs >= 0.5).astype(int)

        test_logloss = log_loss(y_test, test_probs, labels=[0, 1])
        test_accuracy = accuracy_score(y_test, test_preds)

        grad_norm = norm(logistic_grad(w, Xtr_i, y_train))
        hess_cond = hessian_condition(w, Xtr_i)

        practical_success = (
            bool(result.success)
            and np.isfinite(test_logloss)
            and np.isfinite(hess_cond)
            and hess_cond < CONDITION_THRESHOLD
            and np.isfinite(grad_norm)
            and grad_norm < GRADIENT_THRESHOLD
        )

        false_success = bool(result.success) and not practical_success

        return {
            "seed": seed,
            "severity": severity_name,
            "scale_range": f"1e-{scale_exp} to 1e{scale_exp}",
            "scale_exp": scale_exp,
            "method": method,
            "optimizer_success": bool(result.success),
            "practical_success": practical_success,
            "false_success": false_success,
            "test_logloss": test_logloss,
            "test_accuracy": test_accuracy,
            "gradient_norm": grad_norm,
            "hessian_condition": hess_cond,
            "iterations": getattr(result, "nit", np.nan),
            "function_evals": getattr(result, "nfev", np.nan),
            "runtime_seconds": runtime,
            "message": str(result.message)[:120],
        }

    except Exception as e:
        runtime = time.time() - start

        return {
            "seed": seed,
            "severity": severity_name,
            "scale_range": f"1e-{scale_exp} to 1e{scale_exp}",
            "scale_exp": scale_exp,
            "method": method,
            "optimizer_success": False,
            "practical_success": False,
            "false_success": False,
            "test_logloss": np.nan,
            "test_accuracy": np.nan,
            "gradient_norm": np.nan,
            "hessian_condition": np.inf,
            "iterations": np.nan,
            "function_evals": np.nan,
            "runtime_seconds": runtime,
            "message": str(e)[:120],
        }

# ============================================================
# Run severity ablation
# ============================================================

records = []

total = len(SEEDS) * len(SEVERITY_LEVELS) * len(METHODS)
count = 0

for seed in SEEDS:
    for config in SEVERITY_LEVELS:
        for method in METHODS:
            count += 1
            print(
                f"[{count}/{total}] "
                f"seed={seed:03d} | {config['severity']} | "
                f"scale=1e-{config['scale_exp']} to 1e{config['scale_exp']} | "
                f"{method}"
            )

            records.append(
                run_one(
                    seed=seed,
                    severity_name=config["severity"],
                    scale_exp=config["scale_exp"],
                    method=method
                )
            )

raw = pd.DataFrame(records)
raw.to_csv("higgs_severity_ablation_raw_100seeds.csv", index=False)

display(raw.head())

# ============================================================
# Summary by severity and method
# ============================================================

summary_by_method = raw.groupby(["severity", "scale_range", "scale_exp", "method"]).agg(
    runs=("seed", "count"),
    optimizer_success_rate=("optimizer_success", "mean"),
    practical_success_rate=("practical_success", "mean"),
    false_success_rate=("false_success", "mean"),
    mean_test_logloss=("test_logloss", "mean"),
    std_test_logloss=("test_logloss", "std"),
    mean_test_accuracy=("test_accuracy", "mean"),
    mean_gradient_norm=("gradient_norm", "mean"),
    median_gradient_norm=("gradient_norm", "median"),
    median_hessian_condition=("hessian_condition", "median"),
    mean_function_evals=("function_evals", "mean"),
    mean_runtime_seconds=("runtime_seconds", "mean")
).reset_index()

summary_by_method = summary_by_method.sort_values(["scale_exp", "method"])

summary_by_method.to_csv(
    "higgs_severity_ablation_by_method_100seeds.csv",
    index=False
)

display(summary_by_method)

# ============================================================
# Compact paper-ready table:
# collapse across methods to show instability emergence
# ============================================================

paper_summary = raw.groupby(["severity", "scale_range", "scale_exp"]).agg(
    runs=("seed", "count"),
    practical_success_rate=("practical_success", "mean"),
    optimizer_success_rate=("optimizer_success", "mean"),
    false_success_rate=("false_success", "mean"),
    median_hessian_condition=("hessian_condition", "median"),
    mean_gradient_norm=("gradient_norm", "mean"),
    median_gradient_norm=("gradient_norm", "median"),
    mean_function_evals=("function_evals", "mean"),
    mean_runtime_seconds=("runtime_seconds", "mean")
).reset_index()

paper_summary = paper_summary.sort_values("scale_exp")

paper_summary.to_csv(
    "higgs_severity_ablation_paper_summary_100seeds.csv",
    index=False
)

display(paper_summary)

# ============================================================
# LaTeX output for paper
# ============================================================

latex_table = paper_summary[[
    "severity",
    "scale_range",
    "practical_success_rate",
    "optimizer_success_rate",
    "false_success_rate",
    "median_hessian_condition",
    "mean_gradient_norm",
    "mean_function_evals",
    "mean_runtime_seconds"
]].copy()

latex_table.columns = [
    "Severity",
    "Scale Range",
    "Practical Success Rate",
    "Optimizer Success Rate",
    "False Success Rate",
    "Median Hessian Condition",
    "Mean Gradient Norm",
    "Mean Function Evals",
    "Mean Runtime (s)"
]

print("\nPaper-ready table:\n")
display(latex_table)

print("\nLaTeX table:\n")
print(latex_table.to_latex(
    index=False,
    float_format=lambda x: f"{x:.3e}" if abs(x) < 1e-2 or abs(x) > 1e3 else f"{x:.4f}",
    caption="Severity ablation for induced scale imbalance in the HIGGS optimization workflow across 100 randomized train/test splits.",
    label="tab:higgs_severity_ablation"
))

latex_table.to_csv("paper_table_higgs_severity_ablation_100seeds.csv", index=False)

print("Saved:")
print("- higgs_severity_ablation_raw_100seeds.csv")
print("- higgs_severity_ablation_by_method_100seeds.csv")
print("- higgs_severity_ablation_paper_summary_100seeds.csv")
print("- paper_table_higgs_severity_ablation_100seeds.csv")
```
